```python
import os
import sys
import subprocess
import webbrowser
from pathlib import Path

if not main_data_folder or not os.path.isdir(main_data_folder):
    raise ValueError(f"Invalid main_data_folder: {main_data_folder}")

repo_root = Path('/Users/qixinyang/Documents/AdamLab/miniVI_Renana_Pipeline')
app_script = repo_root / 'miniVI_PlaceCell_analysis_V4' / 'dash_cs_detection_app' / 'app.py'

if not app_script.is_file():
    raise FileNotFoundError(f"Cannot find app.py at: {app_script}")

browser_host = '127.0.0.1' if host in ('0.0.0.0', '::', '') else host
url = f'http://{browser_host}:{int(port)}'

# Stop previous GUI process from this kernel if still running
if '_cs_gui_proc' in globals() and _cs_gui_proc is not None and _cs_gui_proc.poll() is None:
    _cs_gui_proc.terminate()

cmd = [
    sys.executable,
    str(app_script),
    str(main_data_folder),
    '--host', str(host),
    '--port', str(int(port)),
    '--no-browser',
]

_cs_gui_proc = subprocess.Popen(
    cmd,
    cwd=str(repo_root / 'miniVI_PlaceCell_analysis_V4'),
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL,
    start_new_session=True,
)

_ = webbrowser.open(url)
```
